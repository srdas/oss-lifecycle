# Generate all developer stats

Collects model data about the developers in all projects.
This gets the following for each project and creates a latex table:
1. p: coefficient of independent propensity to join the project
2. q: coefficient of network intensity to join the project
3. m: calibrated developer-months expected over the project's lifetime
4. t: current life of project in months
5. T: expected lifetime of project in months
6. yrs: expected remaining lifetime of project in years

```python
import os
import numpy as np
import pandas as pd
import sys
sys.path.append('../src')
import fit_bass
import matplotlib.pyplot as plt

# Months between two dates in YYYY-MM-DD str format
def months_between(start_date, end_date):
    start = pd.to_datetime(start_date)
    end = pd.to_datetime(end_date)
    return (end.year - start.year) * 12 + end.month - start.month

os.chdir('../')
```

```python
# Get list of projects for which data files have been collected

project_files = os.listdir("data")
projects_list = []
for file in project_files:
    if file.endswith("-monthly.csv"):
        projects_list.append(file.split("-monthly")[0])

print("Projects:", projects_list)
```

```python
# Collect stats by project
df = pd.DataFrame(columns=["Project", "Start_Date", "End_Date", "p", "q", "m", "t", "T", "yrs"])
for project in projects_list:
    start_date, end_date, p, q, m, t, T, yrs = fit_bass.get_project_developer_model_stats(project)
    df = pd.concat([df, pd.DataFrame([{"Project": project, "Start_Date":start_date, "End_Date":end_date, 
                                       "p": p, "q": q, "m": m, "t": t, "T": T, "yrs": yrs}])], ignore_index=True)
```

```python
# TABLE II
# Export to Latex
df = df.sort_values(by="Project")
print(df.round(5).to_latex(float_format="%.5f", index=False))
```

```python
# Export to CSV
df.to_csv("data/developer_stats.csv", index=False)
df = df.dropna()
```

```python
# FIGURE 5
# Make combined plot of all projects
earliest_start = df['Start_Date'].min() # Get earliest start date of all projects

plt.figure(figsize=(15, 6))
start_datetime = pd.to_datetime(earliest_start)
color_num = 0
for index, row in df.iterrows():
    color_num += 1

    # Get time when series starts in span
    start_offset = months_between(earliest_start, row["Start_Date"])

    # Get the project life cycle so far
    t_list = np.arange(0, int(row["t"]))
    f, _ = fit_bass.bass(row["p"], row["q"], t_list)
    t_list = t_list + start_offset
    datetime_list = [start_datetime + pd.DateOffset(months=offset) for offset in t_list]
    plt.plot(datetime_list, f*row["m"], linewidth=2, label=row["Project"], color=plt.cm.tab20(color_num % 20))

    # Get the project life cycle from now to end
    t_list = np.arange(int(row["t"]), int(row["T"]))
    f, _ = fit_bass.bass(row["p"], row["q"], t_list)
    t_list = t_list + start_offset
    datetime_list = [start_datetime + pd.DateOffset(months=offset) for offset in t_list]
    plt.plot(datetime_list, f*row["m"], linewidth=2, ls='--', color=plt.cm.tab20(color_num % 20))
    plt.legend()
plt.xlabel('Date')
plt.ylabel('No. of Developers making contributions in each month')
plt.title('Developer engagement for selected projects')
plt.grid()
plt.savefig(f"images/all_project_lifecycles.png")
plt.show()
```

```python
# FIGURE 6 (new)
# Make normalized plot of all projects

plt.figure(figsize=(15, 6))
color_num = 0
for index, row in df.iterrows():
    if row["Project"] != "kubeflow-kubeflow": # the peak is at the start (edge case)
        color_num += 1
    
        # Get the peak time
        t0 = np.log(row["q"]/row["p"])/(row["p"]+row["q"])
    
        # Get lifespan
        T = row["T"]
        t_list = np.linspace(-5, T, 100)
        
        # Use Bass
        f, _ = fit_bass.bass(row["p"], row["q"], t_list)
        f0 = (row["p"] + row["q"])**2/(4*row["q"])
        plt.plot(t_list/t0, f/f0, linewidth=2, label=row["Project"], color=plt.cm.tab20(color_num % 20))
    
        plt.legend()
plt.xlabel('Date')
plt.ylabel('No. of Developers making contributions in each month')
plt.title('Developer engagement for selected projects')
plt.grid()
plt.savefig(f"images/all_project_lifecycles_normalized.png")
plt.show()
```

## Plot m vs alpha

$$
\alpha = \ln(q/p)
$$

Note that this fails for $q<p$. 

```python
y = np.log(df["q"]/df["p"])
x = df["m"]

plt.scatter(x,y)
plt.xlabel('Total Dev-months = $m$')
plt.ylabel('$\alpha')
plt.title('Developer engagement for selected projects')
plt.grid()
plt.savefig(f"images/m_alpha.png")
plt.show()

print("Correlation =", np.corrcoef(x,y)[0,1])
```
